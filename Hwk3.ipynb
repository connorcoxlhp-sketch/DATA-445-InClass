```python
# (a) Read the csv data file and create the car_price data-frame
import pandas as pd

car_price = pd.read_csv('CarPrice_Assignment.csv')
car_price.head()
```

```python
# (b) Split the data into train (80%) and test (20%)
from sklearn.model_selection import train_test_split

X = car_price[['wheelbase', 'enginesize', 'compressionratio', 'horsepower', 'peakrpm', 'citympg', 'highwaympg']]
Y = car_price['price']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# (c) LASSO variable selection repeated 1000 times
from sklearn.linear_model import Lasso, LassoCV
from sklearn.model_selection import KFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler


k = KFold(n_splits=5, shuffle=True, random_state=42)

lasso_cv_md = make_pipeline(StandardScaler(), LassoCV(cv=k, n_jobs=-1, random_state=1000))

lasso_cv_md.fit(X, Y)

lasso_md = make_pipeline(StandardScaler(), Lasso(alpha=lasso_cv_md.named_steps["lassocv"].alpha_))

lasso_md.fit(X, Y)

coef_df = pd.DataFrame({'Variable': X.columns, 'Coefficient': lasso_md.named_steps["lasso"].coef_})


print(coef_df)





```

```python
# (d) Linear regression model and RMSE on test dataset
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error

lr_md = make_pipeline(StandardScaler(), LinearRegression())

lr_md.fit(X_train, Y_train)

y_pred = lr_md.predict(X_test)

rmse = root_mean_squared_error(Y_test, y_pred)

print(f'Test RMSE: {rmse:.2f}')
```

```python
# (e) Ridge regression model (optimal lambda selection repeated 100 times) and RMSE on test dataset
from sklearn.linear_model import Ridge, RidgeCV

alphas = [0.001, 0.01, 0.1, 1, 10, 100]

optimal_lambdas = []

for i in range(100):
    ridge_cv = make_pipeline(
        StandardScaler(),
        RidgeCV(alphas=alphas, cv=5))

    ridge_cv.fit(X_train, Y_train)
    optimal_lambdas.append(ridge_cv['ridgecv'].alpha_)

lambda_counts = pd.Series(optimal_lambdas).value_counts()

common_lambda = lambda_counts.idxmax()

print("Most common lambda:", common_lambda)

ridge_md = make_pipeline(StandardScaler(), Ridge(alpha=common_lambda))

ridge_md.fit(X_train, Y_train)

y_pred = ridge_md.predict(X_test)

rmse = root_mean_squared_error(Y_test, y_pred)

print(f"Ridge Test RMSE: {rmse:.2f}")
```

*(f) I would use e becuase the RMSE is lower. This model has lower prediction errors giving us a more accurate car prices.

Use `GridSearchCV` with 5-folds to estimate the optimal `epsilon` from the set [1.1, 1.35, 1.5, 1.75, 2] and `alpha` from the set [0.0001, 0.001, 0.01, 0.1, 1].
```python
# (a) Elastic Net model (optimal alpha and l1_ratio via ElasticNetCV) and RMSE on test dataset
from sklearn.linear_model import ElasticNetCV, ElasticNet
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error
import numpy as np

elastic_net = Pipeline([
    ('scaler', StandardScaler()),
    ('elasticnet', ElasticNetCV(
        l1_ratio=[0.1, 0.3, 0.5, 0.7, 0.9, 0.95, 1],
        cv=5
    ))
])

elastic_net.fit(X_train, Y_train)

print("Best alpha:", (round(elastic_net.named_steps['elasticnet'].alpha_, 3)))
print("Best l1_ratio:", elastic_net.named_steps['elasticnet'].l1_ratio_)

elastic_pred = elastic_net.predict(X_test)

elastic_rmse = np.sqrt(mean_squared_error(Y_test, elastic_pred))

print("Elastic Net RMSE:", elastic_rmse.round(2))
```

```python
# (b) Huber Regression model (optimal epsilon and alpha via GridSearchCV) and RMSE on test dataset
from sklearn.linear_model import HuberRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

huber = Pipeline([
    ('scaler', StandardScaler()),
    ('huber', HuberRegressor())
])

param_grid = {
    'huber__epsilon': [1.1, 1.35, 1.5, 1.75, 2],
    'huber__alpha': [0.0001, 0.001, 0.01, 0.1, 1]
}

huber_grid = GridSearchCV(
    huber,
    param_grid,
    cv=5
)

huber_grid.fit(X_train, Y_train)

print("Best parameters:", huber_grid.best_params_)

huber_pred = huber_grid.predict(X_test)

huber_rmse = np.sqrt(mean_squared_error(Y_test, huber_pred))

print("Huber Regression RMSE:", huber_rmse.round(2))
```

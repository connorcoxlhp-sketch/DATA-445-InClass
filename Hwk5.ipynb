## Exercise 3
(10 points) Suppose you are running `AdaBoost` with 4 training examples. At the start of the current iteration, the four examples have the weights shown in the following table. Another column says if the weak classifier got them correct or incorrect. Determine the new weights for these four examples, and fill in the corresponding entries in the table. Show all your work.

| Observation | Old Weight | Correct? | New Weight |
|:-----------:|:----------:|:--------:|:----------:|
| 1 | 0.16 | Correct   |  |
| 2 | 0.64 | Correct   |  |
| 3 | 0.08 | Incorrect |  |
| 4 | 0.12 | Incorrect |  |


**Work / Answer:** *(write your work here; you may use the code cell below)*

```python
# Exercise 3 work (optional)
import numpy as np

old_weight = np.array([.16, .64, .08, .12])
correct = np.array([True, True, False, False])

error = np.sum(old_weight[~correct])

new_weight = old_weight.copy()
new_weight[correct] *= np.sqrt(error / (1-error))
new_weight[~correct] *= np.sqrt((1 - error) / error)

new_weight = new_weight / np.sum(new_weight)

print(new_weight)
```

## Exercise 10
Consider the `framingham.csv` data file. The dataset is available on the Kaggle website, and it is from an ongoing cardiovascular study on residents of the town of Framingham, Massachusetts. The classification goal is to predict whether the patient has 10-year risk of future coronary heart disease (CHD). The dataset provides the patients' information. It includes over 4,000 records and 15 attributes. Each attribute is a potential risk factor. There are demographic, behavioral and medical risk factors.

- **Demographic**
  - Sex: male or female (Nominal)
  - Age: Age of the patient (Continuous - although the recorded ages have been truncated to whole numbers, the concept of age is continuous)
- **Behavioral**
  - Current Smoker: whether or not the patient is a current smoker (Nominal)
  - Cigs Per Day: the number of cigarettes that the person smoked on average in one day (can be considered continuous as one can have any number of cigarettes, even half a cigarette)
- **Medical (history)**
  - BP Meds: whether or not the patient was on blood pressure medication (Nominal)
  - Prevalent Stroke: whether or not the patient had previously had a stroke (Nominal)
  - Prevalent Hyp: whether or not the patient was hypertensive (Nominal)
  - Diabetes: whether or not the patient had diabetes (Nominal)
- **Medical (current)**
  - Tot Chol: total cholesterol level (Continuous)
  - Sys BP: systolic blood pressure (Continuous)
  - Dia BP: diastolic blood pressure (Continuous)
  - BMI: Body Mass Index (Continuous)
  - Heart Rate: heart rate (Continuous - in medical research, variables such as heart rate, though in fact discrete, are considered continuous because of the large number of possible values)
  - Glucose: glucose level (Continuous)
- **Predict variable (desired target)**
  - 10 year risk of coronary heart disease CHD (binary: "1" means "Yes", "0" means "No")

**(a)** (5 points) Using the `pandas` library, read the csv data file and create a data-frame called `heart`. Remove the observations with missing values.

```python
import pandas as pd

heart = pd.read_csv("framingham.csv")
heart = heart.dropna().reset_index(drop=True)
heart.head()
```

**(b)** (50 points) Based on the different interactions that we have so far with this data file, it seems that `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the most important predictor variables when it comes to predicting the likelihood of `TenYearCHD`. Then do the following:

1. Using `TenYearCHD` as the target variable and `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` as the input variables, create a `StratifiedKFold` cross validation object with 10 folds (shuffled, with a fixed `random_state`) so that every fold keeps the proportion of 0s and 1s of the target.
2. In each fold, train on the training folds and predict the risk of coronary disease for the observations in the held-out fold, for each of the following models:
   - `RandomForestClassifier` (`n_estimators = 500`, `max_depth = 5`)
   - `ExtraTreesClassifier` (`n_estimators = 500`, `max_depth = 5`)
   - `AdaBoostClassifier` (`n_estimators = 500`, `learning_rate = 0.01`, with decision trees of `max_depth = 3` as weak learners)
   - `GradientBoostingClassifier` (`n_estimators = 500`, `max_depth = 3`, `learning_rate = 0.01`)
3. Using 10% as the cutoff value (i.e., predict 1 when the estimated probability of `TenYearCHD = 1` is at least 0.10), compute the accuracy and recall on the held-out fold.
4. Average the accuracy and recall over the 10 folds for each model. What model would you use to predict `TenYearCHD` based on the accuracy and recall?

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import recall_score
from sklearn.metrics import accuracy_score

X = heart[['age', 'totChol', 'sysBP', 'BMI', 'heartRate', 'glucose']]
Y = heart['TenYearCHD']

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_recall = []
et_recall = []
ada_recall = []
gb_recall = []

rf_accuracy = []
et_accuracy = []
ada_accuracy = []
gb_accuracy = []


rf_md = RandomForestClassifier(n_estimators=500, max_depth=5, random_state=42)
et_md = ExtraTreesClassifier(n_estimators=500, max_depth=5, random_state=42)
ada_md = AdaBoostClassifier(estimator=DecisionTreeClassifier(max_depth=3), n_estimators=500, learning_rate=0.01, random_state=42)
gb_md = GradientBoostingClassifier(n_estimators=500, max_depth=3, learning_rate=0.01, random_state=42)

for i, (train_index, test_index) in enumerate(skf.split(X, Y)):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = Y.iloc[train_index], Y.iloc[test_index]

    # Random Forest
    rf_md.fit(X_train, y_train)
    rf_pred_proba = rf_md.predict_proba(X_test)[:, 1]
    rf_pred = (rf_pred_proba >= 0.1).astype(int)
    rf_recall.append(recall_score(y_test, rf_pred))
    rf_accuracy.append(accuracy_score(y_test, rf_pred))

    # Extra Trees
    et_md.fit(X_train, y_train)
    et_pred_proba = et_md.predict_proba(X_test)[:, 1]
    et_pred = (et_pred_proba >= 0.1).astype(int)
    et_recall.append(recall_score(y_test, et_pred))
    et_accuracy.append(accuracy_score(y_test, et_pred))

    # AdaBoost
    ada_md.fit(X_train, y_train)
    ada_pred_proba = ada_md.predict_proba(X_test)[:, 1]
    ada_pred = (ada_pred_proba >= 0.1).astype(int)
    ada_recall.append(recall_score(y_test, ada_pred))
    ada_accuracy.append(accuracy_score(y_test, ada_pred))

    # Gradient Boosting
    gb_md.fit(X_train, y_train)
    gb_pred_proba = gb_md.predict_proba(X_test)[:, 1]
    gb_pred = (gb_pred_proba >= 0.1).astype(int)
    gb_recall.append(recall_score(y_test, gb_pred))
    gb_accuracy.append(accuracy_score(y_test, gb_pred))

average_rf_recall = sum(rf_recall) / len(rf_recall)
average_et_recall = sum(et_recall) / len(et_recall)
average_ada_recall = sum(ada_recall) / len(ada_recall)
average_gb_recall = sum(gb_recall) / len(gb_recall)

average_rf_accuracy = sum(rf_accuracy) / len(rf_accuracy)
average_et_accuracy = sum(et_accuracy) / len(et_accuracy)
average_ada_accuracy = sum(ada_accuracy) / len(ada_accuracy)
average_gb_accuracy = sum(gb_accuracy) / len(gb_accuracy)

print(f'Random Forest Average Recall: {average_rf_recall}, Average Accuracy: {average_rf_accuracy}')
print(f'Extra Trees Average Recall: {average_et_recall}, Average Accuracy: {average_et_accuracy}')
print(f'AdaBoost Average Recall: {average_ada_recall}, Average Accuracy: {average_ada_accuracy}')
print(f'Gradient Boosting Average Recall: {average_gb_recall}, Average Accuracy: {average_gb_accuracy}')
```

**Answer:**: Based on these results I would say it depends on if you're looking for a higher recall or a higher accuracy. I think the best model of the 4 to answer our question would be Random Forest. It's average recall of 82.97% and accuracy of 48.91%, which means Random Forest identifies a high percentage of patients at risk while maintaining reasonable accuracy. Gradient Boosting could also be a good choice it just depends on if you want a slightly higher accuracy and slightly lower recall or vice versa. Since identify the patients who may be at risk is more important I would stick with Random Forest.

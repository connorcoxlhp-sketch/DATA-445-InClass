(6 points) Explain the difference between the validation set approach and $k$-fold cross validation, and describe one advantage of using $k$-fold cross validation over the validation set approach.

Using the 500 observations example from above. Validation set would be like choosing an 80/20 split or 400 observations as the training data and 100 as the testing data. The downside to this is that the results are all dependent on the 100 observations you use to test on. In the k-fold, using the same 500 observations and 10 groups, you get 10 groups of 50 observations. 1 group is used as the testing data and 9 are the training. When on group has been tested they rotate out and go to training data for a new group to step up as the testing and so on through all the groups. k-folds advantage over validation set approach is that since it tests on several subsets of the data that it's results are more reliable.

**(c)** (25 points) Perform a 5-folds cross validation with the goal of measuring the performance, in terms of F1-score, of two competing models:

- Using `age`, `currentSmoker`, `totChol`, `sysBP`, `diaBP`, `BMI`, `heartRate`, and `glucose` as the predictor variables, and `TenYearCHD` as the target variable build a logistic regression model under the 5-folds cross validation framework. Compute and store the F1-score for each iteration.
- Using `age`, `currentSmoker`, `totChol`, `BMI`, `heartRate`, and `glucose` as the predictor variables, and `TenYearCHD` as the target variable build a logistic regression model under the 5-folds cross validation framework. Compute and store the F1-score for each iteration.

Use 25% as threshold to change the likelihoods to labels. Make sure to scale the input variables of both models to 0-1 range (see [`MinMaxScaler`](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html)) before you run the 5-fold cross validation framework. Also, you can use the [`f1_score`](https://scikit-learn.org/stable/modules/generated/sklearn.metrics.f1_score.html) function to compute the F1-score.

```python
# Your code here
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import f1_score

scaler = MinMaxScaler()
heart = heart.dropna()

X = heart[['age', 'currentSmoker', 'totChol', 'sysBP', 'diaBP', 'BMI', 'heartRate', 'glucose']]
X2 = heart[['age', 'currentSmoker', 'totChol', 'BMI', 'heartRate', 'glucose']]
Y = heart['TenYearCHD']

X_scaled = scaler.fit_transform(X)
X2_scaled = scaler.fit_transform(X2)

k = KFold(n_splits=5, shuffle=True, random_state=42)

X_model = []
X2_model = []

for train_index, test_index in k.split(X_scaled):
    
    X_train = X_scaled[train_index]
    X_test = X_scaled[test_index]
    
    Y_train = Y.iloc[train_index]
    Y_test = Y.iloc[test_index]
    model = LogisticRegression()
    model.fit(X_train, Y_train)
    
    probability = model.predict_proba(X_test)[:, 1]
    
    Y_pred = (probability >= 0.25).astype(int)
    
    f1 = f1_score(Y_test, Y_pred)
    
    X_model.append(f1)

print("F1 scores for model with all features:", [round(score, 3) for score in X_model])

for train_index, test_index in k.split(X2_scaled):
    
    X_train = X2_scaled[train_index]
    X_test = X2_scaled[test_index]
    
    Y_train = Y.iloc[train_index]
    Y_test = Y.iloc[test_index]
    model = LogisticRegression()
    model.fit(X_train, Y_train)
    
    probability2 = model.predict_proba(X_test)[:, 1]
    
    predictions = (probability2 >= 0.25).astype(int)
    
    score = f1_score(Y_test, predictions)
    
    X2_model.append(score)

print("F1 scores for model without sysBP and diaBP:", [round(score, 3) for score in X2_model])
```

**(d)** (4 points) Report the average F1-score of each of the models. What model would you use to predict `TenYearCHD`? Explain.

```python
# Your code here
print("Average F1 Model 1:", round(sum(X_model) / len(X_model), 3))
print("Average F1 Model 2:", round(sum(X2_model) / len(X2_model), 3))
```

I would use F1 model 1 because it preformed better when the 'sysBP' and 'diaBP' were included.
